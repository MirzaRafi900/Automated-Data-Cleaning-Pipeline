```python
class CafeDataPipeline:


    def __init__(self,file_path):
        self.file_path = file_path
        self.df = None


    def ingest_data(self):
        self.df = pd.read_csv(self.file_path)
        print(f"📊 Data loaded Suffessfully {self.df.shape}")
        return self.df.head()
```

```python


    def clear_numeric_fields(self):

        if self.df is None:
            return

        target_cols = ["Quantity", "Price Per Unit", "Total Spent"]

        for col in target_cols:
            if col in self.df.columns:
                self.df[col] = pd.to_numeric(self.df[col], errors = "coerce")

        missing_total_mask = self.df["Total Spent"].isna()
        self.df.loc[missing_total_mask, "Total Spent"] = (
            self.df.loc[missing_total_mask, "Quantity"]
            *
            self.df.loc[missing_total_mask, "Price Per Unit"]
        )


        print("🌸 Numeric Cleaning Complete: Text Error corced to calculation")
        return self.df.head(5)
```

```python
def standardize_text_and_dates(self):
    """Target 2: Standardize date types and clean missing placeholders."""
    if self.df is None:
        return

    if "Transaction Date" in self.df.columns:
        self.df["Transaction Date"] = pd.to_datetime(
            self.df["Transaction Date"], errors="coerce"
        )

    text_cols = ["Payment Method", "Item", "Location"]
    for col in text_cols:
        if col in self.df.columns:
            self.df[col] = self.df[col].astype(str).str.strip()
            self.df[col] = self.df[col].replace(
                ["UNKNOWN", "unknown", "nan", "None"], "Missing"
            )

    print("⚙️ Text and Dates Standardized successfully.")
    return self.df.head(5)


# Inject this function into your class
CafeDataPipeline.standardize_text_and_dates = standardize_text_and_dates
```
